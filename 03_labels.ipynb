```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([100,200,310,420])
y = np.array([110,290,300,40])
z = np.array([100,120,300,450])
p = np.array([10,150,200,45])

line_style = dict(marker = "."
                  ,markersize = "30"
                  ,markerfacecolor = "#1f6ab5"
                  ,markeredgecolor = "#17161A"
                  ,linestyle = "solid"
                  ,linewidth = "2")

plt.plot(x,y,color = "red",**line_style)
plt.plot(x,z,color = "blue",**line_style)
plt.plot(x,p,color = "green",**line_style)


plt.title("Class Size",
          fontsize ="20",
          fontfamily ="Arial",
          fontweight = "bold",
          color = "Blue")

plt.xlabel("Students",
           color = "Red",
           fontsize="15",
           fontweight = "bold")

plt.ylabel("Marks",
           color = "Green",
           fontsize="20",
           fontweight = "bold")

# plt.tick_params(axis= "both",color="red")
#changes the color of the tick marks, but not the X/Y-axis numbers (tick labels).

plt.tick_params(axis= "both",colors="red")
#colors (plural) → changes both ticks + their labels.


plt.xticks(x)
#plt.xticks(x) tells Matplotlib where to show the marks/numbers on the X-axis.
#Basically: “Put X-axis ticks at the values stored in x


plt.show()
```
