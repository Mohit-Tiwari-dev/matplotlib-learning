```python
import numpy as np
import matplotlib.pyplot as plt

# Bar Chart = Compares categories of databy representing each category with a bar
```

```python
categories = ["fruits","vegetables","dairy","protein"]
quantity = [1,3,2,4]

plt.bar(categories,quantity,color = "lightgreen")

# For Horizontal bar chart
# plt.barh(categories,quantity,color = "lightgreen")


plt.title("Food Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```
