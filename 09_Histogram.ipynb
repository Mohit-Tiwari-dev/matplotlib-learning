```python
# Histogram = A Visual Representation of distribution of Quantitative data.
             # They Group Values into bins(intervals) 
             # And counts how many fall in each range.
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scores = np.random.normal(loc =80,size =100,scale = 10)
# print(type(scores))
# print(scores)


scores =np.clip(scores,0,100)
# Below 0 → becomes 0
# Above 100 → becomes 100
# Between 0–100 → stays unchanged.


plt.hist(scores,bins = 10,
         color = "lightblue",
         edgecolor = "black")

plt.title("Exam Scores")
plt.xlabel("Marks")
plt.ylabel("No of Students")

plt.show()
```
