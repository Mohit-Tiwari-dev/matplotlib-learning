```python
# Figure = The Entire Canvas
# Ax =a single plot(subplot) within a figure
```

```python
import numpy as np
import matplotlib.pyplot as plt

print(plt.subplots(2,2))
# returns a tuple with 2 entities 
# first is a figure object which is a canvas 
# And 2-D array of Axes objects
# Axes is techinically a numpy array
```

```python
figure,axes =plt.subplots(2,2)

# 2,1 -> 2 rows & 1 Column
# 1,2 -> 1 rows & 2 Column

x = np.array([1,2,3,4,5])
axes[0,0].plot(x,x*2,color = "red")
axes[0,0].set_title("x*2")

axes[0,1].plot(x,x**2,color = "red")
axes[0,1].set_title("x**2")

axes[1,0].plot(x,x**3,color = "red")
axes[1,0].set_title("x**3")

axes[1,1].plot(x,x**4,color = "red")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```
