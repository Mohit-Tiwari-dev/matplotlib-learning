```python
import numpy as np
import matplotlib.pyplot as plt

# Pie Chart = Circular Chart divided into slices to show some percentages of total.
#             Good for Visualizing distribtion among categories.
```

```python
categories = ["Btech","BCA","MBA","MCA"]
values = np.array([300,200,150,250])

fav_col =["red","green","blue","yellow"]


plt.pie(values,labels=categories,
        autopct="%1.1f%%",
        colors=fav_col,
        explode = [0,0,0,0.1],
        shadow = True)

# autopct="%1.1f%%" is used in a pie chart to show the percentage on each slice.
# %1.1f → show the number with 1 decimal place
# %% → display the % symbol
plt.title("Pie Chart")
plt.show()
```
