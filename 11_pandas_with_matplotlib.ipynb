```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Pokemon.csv")
# print(df["Type1"].value_counts())

type_counts = df["Type1"].value_counts(ascending=True)

plt.barh(type_counts.index,type_counts.values,color ="Blue",edgecolor = "black")

plt.title("No of Pokemon by Primary Type")
plt.xlabel("Count")
plt.ylabel("Type")
plt.tight_layout()
plt.show()
```
