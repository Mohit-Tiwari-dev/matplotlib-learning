```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([10,5,20,15])
y = np.array([10,40,50,30])


plt.plot(x,y,marker = "."
         ,markersize = "30"
         ,markerfacecolor = "#1f6ab5"
         ,markeredgecolor = "#17161A"
         ,linestyle = "dashdot"
         ,linewidth = "2"
         ,color ="red")

# Shortform ms,mfc,mec,ls,lw
#'solid' A continuous line (Default)
# 'dashed'		A dashed line
# 'dashdot' Alternating dashes and dots
# 'dotted'	A dotted line
# 'None'	Draws no line (useful if you only want markers)

plt.show()
```

```python
x = np.array([100,200,310,420])
y = np.array([110,290,300,40])
z = np.array([100,120,300,450])
p = np.array([10,150,200,45])

line_style = dict(marker = "."
                  ,markersize = "30"
                  ,markerfacecolor = "#1f6ab5"
                  ,markeredgecolor = "#17161A"
                  ,linestyle = "solid"
                  ,linewidth = "2")

plt.plot(x,y,color = "red",**line_style)
plt.plot(x,z,color = "blue",**line_style)
plt.plot(x,p,color = "green",**line_style)

# **line_style -> Unpack the line_style dictionary
plt.show()
```
